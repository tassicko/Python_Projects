```python
import numpy as np

arr_0d = np.array(25)
print('0-D:', arr_0d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print('2-D:\n', arr_2d)

arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print('3-D:\n', arr_3d)
```

```python
# Creating random arrays
arr = np.array([1, 2, 3, 4])
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((3, 2))
random = np.random.rand(2, 2, 2)
arange_arr = np.arange(0, 12, 2)
linspace_arr = np.linspace(0, 1, 5)
copy_arr = random.copy()


print('array:\n', arr)
print('Zeros:\n', zeros_arr)
print('Ones:\n', ones_arr)
print('Arange:\n', arange_arr)
print('Linspace:', linspace_arr)

print(f"3D Array with Random Numbers (2x2x2):\n{random}")
print('\nCopy:\n', copy_arr)
```
