# Making the loop for whole database

```python
#empty dataFrame:
pf = pd.DataFrame()
```

```python
for i in range (500, 530):
    params = {
    'api_key':'46031462051e7fa72f769263a1b7e876',
    'language':'en_US',
    'page':i
    }
    headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"
    }
    response = requests.get("https://api.themoviedb.org/3/movie/top_rated", params= params, headers=headers)

    data = response.json()
    if "results" in data:
        temp_df = pd.DataFrame(data["results"])
        temp_df = temp_df.drop(columns=['backdrop_path', 'genre_ids', 'poster_path'])
        pf = pd.concat([pf, temp_df], ignore_index=True)
    else:
        print("No results on page", i, ":", data)
```
