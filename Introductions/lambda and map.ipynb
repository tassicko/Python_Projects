```python
from anyio.functools import reduce

square = lambda x:x**2
```

```python
print(square)
```

```python
for i in range(1,6):
    print(square(i))
```

```python
number = [1,2,3,4,5,6]
print(list(map(square, number)))
```

```python
print(list(filter(square, number)))
```

```python
even = list(map(lambda x: x % 2 == 0, number))
even
```

```python
even = list(filter(lambda x: x % 2 == 0, number))
even
```
