```python
plt.style.use('default')   # reset
plt.style.use('seaborn-v0_8-pastel')   # apply classic


plt.figure(
    num="TEST FIGURE",        # identifier/name for the figure
    figsize=(7,7),     # (width, height) in inches
    dpi=100,         # resolution: dots per inch
    #facecolor="silver",   # background color of the figure
    #edgecolor="green",   # border color
    frameon=True,     # whether to draw the figure frame
    clear=False,       # if True and figure exists, clear it instead of returning it
    linewidth=1
)

plt.title(
    "Daily Sales",
    fontsize=16,
    fontweight="bold",
    loc="left",
    pad=10,
    rotation = 45
)
plt.xlabel(
    "X axis",
    color='red',
    fontsize=16,
    fontweight="bold",
    loc="left",
    rotation=45,
    alpha=0.85
)
plt.ylabel(
    "Y axis",
    color='green',
    fontsize=16,
    fontweight="bold",
    loc="center",
    rotation=45,
    alpha=0.75
)

plt.plot(data['No'], data['Height'], label= 2022)

plt.legend(loc='best') #works if the .plot() has label
plt.grid(linestyle='-', alpha=0.5)
plt.show()
plt.style.available
```

```python
figure = dict(
    num="TEST FIGURE",        # identifier/name for the figure
    figsize=(7,7),     # (width, height) in inches
    dpi=100,         # resolution: dots per inch
    facecolor="silver",   # background color of the figure
    edgecolor="green",   # border color
    frameon=True,     # whether to draw the figure frame
    clear=False       # if True and figure exists, clear it instead of returning it
)

plt.figure(**figure)
plt.plot([1, 2, 3], [1, 4, 9])
plt.title("Figure 1")

plt.figure(**figure)
plt.plot([1, 2, 3], [9, 4, 1])
plt.title("Figure 2")

plt.show()
```

```python
subplt = {
    "nrows":2,  # number of rows
    "ncols":2,  # number of columns
    "figsize":(5,5),  # (width, height) in inches — same as plt.figure()
    "dpi":100,  # resolution
    "sharex":False,  # share x-axis across subplots
    "sharey":False,  # share y-axis across subplots
    "squeeze":True  # if True, removes extra dimensions when nrows/ncols = 1
}
```

```python
fig, ax = plt.subplots(**subplt)

ax[0, 0].plot([1, 2, 3], [1, 2, 3])
ax[0, 0].set_title("Plot 1")

ax[0, 1].plot([1, 2, 3], [3, 2, 1])
ax[0, 1].set_title("Plot 2")

ax[1, 0].plot([1, 2, 3], [1, 4, 9])
ax[1, 0].set_title("Plot 3")

ax[1, 1].plot([1, 2, 3], [9, 4, 1])
ax[1, 1].set_title("Plot 4")

plt.tight_layout()
plt.show()
```

```python
plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [1, 2, 3])
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.plot([1, 2, 3], [3, 2, 1])
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.plot([1, 2, 3], [1, 4, 9])
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.plot([1, 2, 3], [9, 4, 1])
plt.title("Plot 4")

plt.tight_layout()#this is best for this scenario
plt.show()
```

- **PLOT TYPES**

```python
lineplt = dict(
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=6,
    markerfacecolor="red",
    markeredgecolor="green",
)
```

```python
plt.figure(**figure)
plt.plot(data['No'], data['Height'], **lineplt)

plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Daily Sales")

plt.minorticks_on()
plt.tight_layout()
plt.show()
```
