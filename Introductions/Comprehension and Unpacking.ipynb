- **LIST COMPREHENSION**

```python
number = []

for i in range(5):
    number.append(i)
print(number)
```

```python
number = [i for i in range(1,5)]
number
```

```python
squares = [x**2 for x in range(2, 10)]
squares
```

- **Conditional Comprehension**

```python
power = [2**x for x in range(1,10)
            if x%2==0]
power
```

```python
even_squares = {
    x: x ** 2
    for x in range(1,10)
    if x % 2 == 0
}
even_squares
```

- **'*'and ** Unpacking**

```python
number = [1,2,3,4,5]
print(*number)
print(number)
```

```python
person1 = ['Tasik', 'islam']
person2 = ['Tabassum', 'Mugdha']

names = [*person2, *person1]
print(*names)
```

- **Unpacking Dictionary**

```python
def show(Name, Section):
    print(Name)
    print(Section)

person = { 'Name': ["Tasik", "Mugdha"],
           'Section': ["A", "A"]
           }
show(**person)
```

```python
person = { 'Name': "Tasik",
           'Section': "A"
           }
print(**person)
```

```python
numbers = [10, 20, 30, 40, 50, 60]

first, *middle, q, l = numbers
print(f" {first}\n {middle}\n {l}\n {q}")
```
