**🔑 Concept**
- A container is any Python object that can hold other objects.

- An iterable is any Python object you can loop over with for ... in ....

- Not all data types are containers, but all containers are iterables (if they implement __iter__ or __getitem__).

```python
# Iterable examples
for ch in "Python":
    print(ch, sep='', end='') # P y t h o n
print()
for x in ['a','b','c','d','e']:
    print(x, sep="", end="")   # 1 2 3
print()

for x in {1,2,3,4}:
    print(x, sep="", end="")
print()

for k in {"a":1, "b":2}:
    print(k, sep="", end="")   # the key is default iterable
print()
```
