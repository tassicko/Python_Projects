- **Checking if int, bool, string are immutable or not**

```python
name = "python"
print(id(name))
```

```python
print(id(name.upper()))
print(id(name))
print(('no','yes')[id(name.upper()) == id(name)])
```

```python
name = 'Python'
print(id(name))
```

```python
name is name.upper()
```

```python
x = [1,2,3]
y=x
print(x is y)
print(f"{id(x)}\n{id(y)}")
```

```python
values = [0, "\n", [], "Python", 10]

for value in values:
    if value:
        print(f"True: {value}")
    else:
        print(f"False: {value}")
```
