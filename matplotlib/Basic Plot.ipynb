## Excersice 4

- # How to use subplot() and subplots()
-  plt.subplot()
- fig, ax = plt.subplot()

```python
p = np.array([10,20,30])
q = np.array([20,40,10])
r = np.array([40,10,30])

pq = pd.DataFrame(q,p)
pr = pd.DataFrame(r,p)

fig, ax = plt.subplots()
pq.plot(ax=ax, label='line1')
pr.plot(ax=ax, label='line2')

plt.axis([10,30,10,40])
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Row 1, Col 2, Position 1
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [4, 5, 6])
plt.axis([1,3,4,6])
plt.title("First Plot")

# Row 1, Col 2, Position 2
plt.subplot(1, 2, 2)
plt.plot([1, 2, 3], [6, 5, 4])
plt.title("Second Plot")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Creates the layout immediately
fig, axes = plt.subplots(1, 2)

# Access by index (0-based)
axes[0].plot([1, 2, 3], [4, 5, 6])
axes[0].set_title("First Plot")

axes[1].plot([1, 2, 3], [6, 5, 4])
axes[1].set_title("Second Plot")

plt.tight_layout()
plt.show()
```

## Excercise 5
- plt.plot(x, y, color='red', linestyle='dashdot', linewidth = 3,
         marker='o', markerfacecolor='blue', markersize=12)

```python
import matplotlib.pyplot as plt
# x axis values
x = [1,4,5,6,7]
# y axis values
y = [2,6,3,6,3]

# plotting the points
plt.plot(x, y, color='red', linestyle='dashdot', linewidth = 3,
         marker='o', markerfacecolor='blue', markersize=12)
#Set the y-limits of the current axes.
plt.ylim(1,8)
#Set the x-limits of the current axes.
plt.xlim(1,8)

# naming the x axis
plt.xlabel('x - axis')
# naming the y axis
plt.ylabel('y - axis')

# giving a title to my graph
plt.title('Display marker')
# function to show the plot
plt.show()
```

## Excercise 6
 - plt.axis([0, 100, 0, 200])


```python
import matplotlib.pyplot as plt
X = range(1, 50)
Y = [value * 3 for value in X]
plt.plot(X, Y)
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Draw a line.')
# shows the current axis limits values
print(plt.axis())
# set new axes limits
# Limit of x axis 0 to 100
# Limit of y axis 0 to 200
plt.axis([0, 100, 0, 200])
# Display the figure.
plt.show()
```

## multiple plot in one line

```python
import numpy as np
import matplotlib.pyplot as plt

# Sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)

# green dashes, blue squares and red triangles
plt.plot(t, t, 'g--', t, t**2, 'bs', t, t**3, 'r^')
plt.show()
```
