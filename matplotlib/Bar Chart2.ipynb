```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

 ## Exercise 1: plt.grid
- plt.minorticks_on()
- plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
- plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')


```python
language = np.array(['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++'])
popularity = np.array([22.2, 17.6, 8.8, 8, 7.7, 6.7])
df = pd.DataFrame({
#    'language': language,
    'popularity': popularity
}).set_index(language)
df
df.plot()
```

```python
plt.figure(facecolor='black')
plt.bar(language, popularity, color='blue', edgecolor='black')
plt.style.use('ggplot')


plt.xlabel("Language", color='white')
plt.ylabel("Popularity", color='white')
plt.title("""Popularity of Programming Language
          "Worldwide, Oct2017 compared to a year ago""", color='white')
plt.xticks(color='white')
plt.yticks(color='white')

plt.ylim([0,25])
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
```

```python
plt.figure(facecolor='black')
plt.barh(language, popularity, color='green', edgecolor='black', alpha=0.75)
plt.style.use('ggplot')


plt.xlabel("Language", color='white')
plt.ylabel("Popularity", color='white')
plt.title("""Popularity of Programming Language
          "Worldwide, Oct2017 compared to a year ago""", color='white')
plt.xticks(color='white')
plt.yticks(color='white')

plt.xlim([0,25])
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
```

```python
plt.figure(facecolor='black')
plt.bar(language, popularity, color='blue', edgecolor='black')
plt.style.use('ggplot')


plt.xlabel("Language", color='white')
plt.ylabel("Popularity", color='white')
plt.title("""Popularity of Programming Language
          "Worldwide, Oct2017 compared to a year ago""", color='white')
plt.xticks(color='white')
plt.yticks(color='white')

plt.ylim([0,25])

plt.subplots_adjust(bottom=1, top=2.2)

plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
```
