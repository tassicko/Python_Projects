- **pyplot is a module that is user friendly interface for ploting**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25 , 30 , 20])
y2 = np.array([17, 18, 38, 20])
y3 = np.array([10, 20, 30, 40])
```

- **#making dictionary**

```python
line_style= dict(
marker="v",
markersize= 10,
markerfacecolor="red",
markeredgecolor="green",
linestyle="solid",
linewidth=4,
)

title_style= dict(
fontsize=20,
family="Arial",
fontweight="bold",
)
```

-

-

-

-

-

-

```python
try:
    plt.plot(x,y1)
    plt.show()
except ValueError as error:
    print(error)
```

```python
plt.plot(x,y2,
         marker="v",
         markersize= 10,
         markerfacecolor="red",
         markeredgecolor="green",
         linestyle="solid",
         linewidth=4,
         color="purple"
)
plt.show()
```

-

-

- **change tik values in x and y axis**

```python
plt.xticks(x)
plt.yticks(y3)

plt.tick_params(axis="both",
                colors="blue")
plt.show()
```

-

-

- **Unpack the dictionary method**

```python
try:
    plt.plot(x,y2, color="yellow",**line_style)
    plt.plot(x,y3,color="magenta",**line_style)
    plt.show()
except (TypeError, NameError) as error:
    print(error)
```

- **USING LABELING AND TITLES**

```python
plt.xlabel("Year",
           fontsize=20,
           family="Arial",
           fontweight="bold",
           color="green")

plt.ylabel("Students",
           fontsize=20,
           family="Arial",
           fontweight="bold",
           color="green")

plt.title("Class Size",
          fontsize=20,
          family="Arial",
          fontweight="bold",
          color="pink")
plt.plot(x,y1, **line_style, color= "red")
plt.plot(x,y2, **line_style, color= "blue")
plt.plot(x,y3, **line_style, color= "magenta")
plt.show()
```

```python
plt.xlabel("Year", **title_style, color="red")

plt.ylabel("Students", **title_style, color="blue")

plt.title("Class Size", **title_style, color="green")

plt.plot(x,y1, **line_style, color= "red")
plt.plot(x,y2, **line_style, color= "blue")
plt.plot(x,y3, **line_style, color= "magenta")

plt.xticks(x)
plt.yticks(y3)
plt.show()
```

- **Make grid lines using grid() function**

```python
plt.grid()
plt.plot(x,y2)
plt.show()
```

```python
#grid funtion customization
try:
    plt.grid(axis="y",
             linewidth= 2,
             color="lightgray",
             linestyle="dashed",
             )
except ValueError as error:
    print(error)
```

- **Bar Chart**

```python
categories = np.array(["Grains", "Fruits", "Vegetables", "Protien", "Dairy", "Sweets"])
values = np.array([4,5,6,7,7,8])
```

```python
plt.bar(categories, values)#check all the bar() parameters like color
plt.show()
```

```python
plt.barh(categories, values)
plt.show()
```

- **Pie Chart**

```python

plt.pie(values, labels=categories,#autopct = automatic percentage
                autopct=f"%1.1f%%",
                colors=["red", "blue", "yellow", "green"],
                explode=[0,0,0,0.1,0,0],   # make a slice explode
                shadow=True,
                startangle=45
        )
plt.show()
```

```python

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 1].plot(x, y2)
axes[1, 0].plot(x, y3)
axes[1, 1].plot(x, y1*2)

plt.show()
```
