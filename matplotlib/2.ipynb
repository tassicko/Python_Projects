```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Z:/codes/python/datasets/titanictrain.csv")
df
```

```python
a = df.groupby("Sex")["Age"].count()
b = pd.DataFrame(a).reset_index()
b
list(b['Sex'])
```

```python
plt.style.available
```

```python
plt.style.use("fast")
plt.figure(
    dpi=100,
    figsize=(15,10),
    frameon=True
)
x = np.array(b['Sex'])
y = np.array(b['Age'])

plt.minorticks_on()
plt.bar(x,y)
plt.show()
```

```python
# 1. Define positions
x = [1, 2, 3, 4, 5,6,7,8]
y = [10, 14, 12, 18, 15,20,22,24]

# 2. Define numerical values for colors (e.g., Intensity or Temperature)
# The lowest number (10) gets the lowest color on the gradient.
# The highest number (50) gets the highest color.
intensities = [10, 15, 20, 25, 30,35, 40, 50]

# 3. Plot using c and cmap together
plt.scatter(x, y,c=intensities, vmin=10, vmax=50, cmap='coolwarm', s=100, edgecolors='black')

# 4. Add the color key guide
plt.colorbar(label='Intensity Level')

plt.show()
```
